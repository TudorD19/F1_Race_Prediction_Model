```python
import sys
sys.path.append("../src")

import pandas as pd
from openf1_client import get
from lap_table import build_lap_table

fp2 = get("sessions", use_cache=False, year=2026, session_name="Practice 2")
races = get("sessions", use_cache=False, year=2026, session_name="Race")
sessions = pd.concat([fp2, races])
```

```python
tables = []
for _, s in sessions.iterrows():
    t = build_lap_table(s["session_key"])
    if t is not None:
        t["session_name"] = s["session_name"]
        t["location"] = s["location"]
        tables.append(t)

all_laps = pd.concat(tables, ignore_index=True)
all_laps.to_csv("../data/processed/laps_2026.csv", index=False)

all_laps.groupby("session_name").size()
```
