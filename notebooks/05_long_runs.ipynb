```python
import pandas as pd

laps = pd.read_csv("../data/processed/laps_2026.csv")
fp2 = laps[laps["session_name"] == "Practice 2"].copy()

fp2["laps_in_stint"] = fp2.groupby(["session_key", "driver_number", "stint_number"])["lap_number"].transform("size")

fp2[["location", "driver_number", "stint_number", "lap_number", "laps_in_stint"]].head(15)
is_long_run = fp2["laps_in_stint"] >= 5
long_runs = fp2[is_long_run].copy()


print(len(long_runs))                                                       # 1920 laps
print(long_runs.groupby(["session_key", "driver_number", "stint_number"]).ngroups)  # 229 long runs
print(long_runs.groupby(["session_key", "driver_number"]).ngroups)         # 213 driver-weekends

drivers_per_session = long_runs.groupby("session_key")["driver_number"].transform("nunique")
min_drivers = drivers_per_session >= 8

long_runs = long_runs[min_drivers].copy()
long_runs.to_csv("../data/processed/long_runs_2026.csv", index=False)
```
